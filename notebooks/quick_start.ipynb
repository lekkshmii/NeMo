# Quick Start: NeMo Financial Research Assistant

## 🚀 Get up and running with NeMo in 5 minutes

```python
import sys
sys.path.append('../src')

from ingestion.document_processor import DocumentProcessor
from extraction.financial_extractor import FinancialExtractor
from embeddings.vector_store import VectorStore
from reasoning.query_engine import QueryEngine
from data_sources.sec_api import SECDataSource

nemo = {
    'processor': DocumentProcessor(),
    'extractor': FinancialExtractor(),
    'vector_store': VectorStore(),
    'query_engine': QueryEngine(),
    'sec_api': SECDataSource()
}

print("🧠 NeMo Financial Research Assistant loaded!")
```

## Example 1: Analyze Apple's Latest 10-K

```python
apple_info = nemo['sec_api'].get_company_info_by_ticker("AAPL")
print(f"Company: {apple_info['name']} (CIK: {apple_info['cik']})")

filings = nemo['sec_api'].get_company_filings(apple_info['cik'], ['10-K'], 1)
latest_10k = filings[0]
print(f"Latest 10-K: {latest_10k['filing_date']} - {latest_10k['accession_number']}")
```

## Example 2: Ask Questions About Financial Documents

```python
questions = [
    "What was Apple's revenue growth?",
    "What are the main risk factors?",
    "How much did Apple spend on R&D?"
]

for question in questions:
    search_results = nemo['vector_store'].search(question, n_results=3)
    answer = nemo['query_engine'].answer_query(question, search_results, "local")
    
    print(f"Q: {question}")
    print(f"A: {answer['response'][:200]}...")
    print()
```

## Example 3: Real-time Market Monitoring

```python
recent_filings = nemo['sec_api'].search_recent_filings("8-K", 2)
print(f"Found {len(recent_filings)} recent 8-K filings:")

for filing in recent_filings[:5]:
    print(f"- {filing['company_name']}: {filing['date_filed']}")
```
